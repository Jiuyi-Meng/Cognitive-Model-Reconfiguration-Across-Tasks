```python
import torch

from assignment import generate_dataset, N_TRIALS_TEST
from myRNN import TaskRNN


# --------------------
# 1. Generate new test data
# --------------------

X_test, Y_test = generate_dataset(N_TRIALS_TEST)

X_test = torch.tensor(X_test)
Y_test = torch.tensor(Y_test)


# --------------------
# 2. Create model
# --------------------

model = TaskRNN(
    input_size=12,
    hidden_size=64,
    output_size=5
)


# --------------------
# 3. Load trained weights
# --------------------

model.load_state_dict(
    torch.load("task_rnn.pth")
)


# --------------------
# 4. Testing mode
# --------------------

model.eval()


# --------------------
# 5. Run test
# --------------------

with torch.no_grad():

    output, hidden_activity = model(X_test)

    predicted = torch.argmax(
        output,
        dim=1
    )

    accuracy = (
        predicted == Y_test
    ).float().mean()


# --------------------
# 6. Results
# --------------------

print(
    f"Test Accuracy: {accuracy.item():.4f}"
)

print(
    "Hidden activity shape:",
    hidden_activity.shape
)
```

```python
# --------------------
# Separate two tasks
# --------------------

task = torch.argmax(
    X_test[:, 0, 10:12],
    dim=1
)

color_trials = task == 0
shape_trials = task == 1


color_accuracy = (
    predicted[color_trials] == Y_test[color_trials]
).float().mean()

shape_accuracy = (
    predicted[shape_trials] == Y_test[shape_trials]
).float().mean()


print(
    f"Color Task Accuracy: {color_accuracy.item():.4f}"
)

print(
    f"Shape Task Accuracy: {shape_accuracy.item():.4f}"
)
```
